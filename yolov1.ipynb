```python
from ultralytics import YOLO

model = YOLO('yolo11n.pt') 

# Nhớ sửa lại đoạn [TEN_DATASET] cho khớp với tên bạn vừa đặt
model.train(
    data='/kaggle/input/datasets/linhlthuy/food-ingredient-recognition-v4i-yolov11/Food Ingredient Recognition.v4i.yolov11/data.yaml', 
    epochs=100, 
    imgsz=640,
    batch=64  
)
```
